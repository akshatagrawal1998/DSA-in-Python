## List vs Arrays in Python

````python
import sys

arr = []

for i in range(10):
    arr.append(i)
    print(f"Length: {len(arr)}, Size in memory: {sys.getsizeof(arr)}")
````

## Above observation:

Memory does NOT increase every time
- It increases in chunks

### Why?

Python internally:

• allocates extra memory

• avoids resizing every time

This is called dynamic array with over-allocation

### Python lists are implemented as dynamic arrays, which:

- automatically resize when capacity is exceeded
- use extra memory allocation to reduce resizing cost
- support efficient append operations

### Internal Working: Behind the scenes:

- Initial small memory allocated
- When full → new bigger array created
- Old elements copied
- New space used

# **Linear Search**

### Linear Search is a searching algorithm where we check each element one by one until we find the target element or reach the end of the list.

### Suppose you have 10 books on a table.
You want to find ‘Python Book’.
What will you do?
You will check each book one by one until you find it.”

-> That is Linear Search

## Algorithm Steps:
1. Start from the first element
2. Compare with target
3. If match → return index
4. Else → move to next element
5. Repeat till end
6. If not found → return -1

## Pseudo Code:

START

FOR each element in the list:

    IF element == target:

        RETURN index
RETURN -1

END

````python
def linear_search(arr, target):
    for i in range(len(arr)):
        if arr[i] == target:
            return i
    return -1

arr = [10, 20, 30, 40, 50]
target = 30

linear_search(arr, target)
````

## TIME COMPLEXITY:

### Best Case: Element found at first position: O(1)

###Worst Case: Element at last OR not present O(n)

###Average Case: O(n)

## Usecases:
1. Small datasets
Example: Searching contacts in a small list

2. Unsorted data
Binary search won't work

3. Real-time scanning
Checking logs, streams, etc.

````python
# alternate

def linear_search_pythonic(arr, target):
    if target in arr:
        return arr.index(target)
    return -1

arr = [10, 20, 30, 40, 50]
target = 30

linear_search_pythonic(arr, target)
````

# **Binary Search**
https://leetcode.com/problems/binary-search/description/
### Problem Binary Search Solves

You have a sorted dataset, and you want to find something fast.

Instead of checking one-by-one (linear search), we

- Eliminate half of the data every step



## Algorithm (Step-by-Step)

1 Initialize:
left = 0, right = n-1

2 While left ≤ right:

Find mid
Compare arr[mid] with target

3 If equal → return index

4 If target > arr[mid] → search right

5 If target < arr[mid] → search left

### When Array is NOT Sorted, can we apply binary search?
Answer: NO
In that case:
- Sort first → O(n log n)
- Or use linear search → O(n)

### If there are duplicates in array: which index will you get?
Answer: Any one (not guaranteed)

### Time Complexity of Binary Search

Best Case → O(1)

Worst Case → O(log n)

### Space Complexity

Iterative → O(1)

Recursive → O(log n)

````python
# binary search - O(logn)
## https://leetcode.com/problems/binary-search/description/

def binary_search(arr, target):

    left = 0 # index of first element
    right = len(arr) - 1

    while left <= right:

        mid = (left + right) // 2

        if arr[mid] == target:
          return mid # target is found at the mid position, so return that index itself.

        elif arr[mid] < target: # target lies on the right side
          left = mid + 1 ## search in the right half of the array

        else:
          right = mid - 1 # search in the left half of the array

    return -1 # element not found


arr = [1,2,4,6,7,8,9,9]
target = 9

binary_search(arr, target)
````

# **Lower Bound** -
https://leetcode.com/problems/search-insert-position/description/
### Index of the first element equal to or greater than the target in a sorted array

Given a sorted array of distinct integers and a target value, return the index if the target is found. If not, return the index where it would be if it were inserted in order.


Find the smallest index such that value ≥ target

### Real-Life Use Cases
1. E-commerce (Amazon/Flipkart): Find first product ≥ ₹1000

2. Data Analytics: Find first date where sales ≥ target

3. Student Marks: Find first student who scored ≥ passing marks

4. Ride Pricing: Find first surge price ≥ user budget

````python
#lower bound code / search insert position

def lower_bound(arr, target):

    left, right = 0, len(arr)-1
    ans = len(arr) # if the element is larger than all the elements present, then ans = last index
    while left <= right:
      m = (left+right)//2
      if arr[m] >= target:
        ans = m
        right = m-1
      else:
        left = m+1
    return ans

arr = [1,2,4,6,7,8,9,90]
target = 5

lower_bound(arr, target)
````

````python
# alternate code

def lower_bound(arr, target):

    left, right = 0, len(arr)

    while left < right:

        mid = (left + right)//2

        if arr[mid] < target:
            left = mid + 1
        else:
            right = mid

    return left


arr = [1,2,4,6,7,8,9,90]
target = 9

lower_bound(arr, target)
````

# **Search in rotated array**
https://leetcode.com/problems/search-in-rotated-sorted-array/description/

## e.g.: [4, 5, 6, 7, 1, 2, 3]

### Algorithm (Step-by-Step)

1 Initialize:

left = 0, right = n-1

2 While left ≤ right:

Find mid
If arr[mid] == target → return

3 Check:

Case 1: Left half is sorted
arr[left] <= arr[mid]

Now check:

if target is between arr[left] and arr[mid]
    → go left
else
    → go right

Case 2: Right half is sorted
else

Now check:

if target is between arr[mid] and arr[right]
    → go right
else
    → go left

  

````python
def search_rotated(arr, target): # arr is rotated sorted array

    left, right = 0, len(arr) - 1

    while left <= right:

        mid = (left + right) // 2

        # Found target
        if arr[mid] == target:
            return mid

        # Left half is sorted
        if arr[left] <= arr[mid]:

            if arr[left] <= target < arr[mid]:
                right = mid - 1
            else:
                left = mid + 1

        # Right half is sorted
        else:

            if arr[mid] < target <= arr[right]:
                left = mid + 1
            else:
                right = mid - 1

    return -1

arr = [4,5,6,7,0,1,2]
target = 5

search_rotated(arr, target)
````

# **Index of Peak element in Mountain array**

https://leetcode.com/problems/peak-index-in-a-mountain-array/description/

# 1. Problem Statement (Crystal Clear)

## Mountain Array Definition

An array is called a **mountain array** if:

1 It strictly increases
2 Then strictly decreases
3 Has exactly one peak

Example:

```text
[1, 3, 5, 7, 6, 4, 2]
         ↑
       Peak = 7


# 1. Brute Force Approach (Start Simple)

for i in range(1, len(arr)-1):
    if arr[i] > arr[i-1] and arr[i] > arr[i+1]:
        return i

## Complexity: O(n)

## Observation

At any index `mid`, compare:
arr[mid] and arr[mid+1]

### Case 1: Increasing Slope

arr[mid] < arr[mid+1]

Example:
3 < 5

This means:

→ You are on the **left side of peak**
→ Peak lies on the **right side**

So:

left = mid + 1

### Case 2: Decreasing Slope

arr[mid] > arr[mid+1]

Example:

7 > 6

This means:

→ You are on the **right side of peak** OR at peak
→ Peak lies on the **left side (including mid)**

So:
right = mid

# 5. Algorithm (Step-by-Step)

1 Initialize:
left = 0
right = n - 1

2 While left < right:

* Find mid
* Compare arr[mid] with arr[mid+1]

3 If increasing → move right
4 If decreasing → move left

5 When left == right → that is peak


# 9. Time and Space Complexity

Time: O(log n)

Space: O(1)

````python
# brute force approach

arr = [1, 3, 5, 7, 6, 4, 2]

for i in range(1, len(arr)-1):  # iterating over each item:
  if arr[i] > arr[i-1] and arr[i] > arr[i+1]:
    print(i)

````

````python
# alternate code (better):

def find_peak_mountain(arr):

    left = 0
    right = len(arr) - 1

    while left < right:

        mid = (left + right) // 2

        if arr[mid] < arr[mid + 1]: # increasing slope
            left = mid + 1 # go to the right side
        else:
            right = mid # go to the left side

    return left   # index of peak


arr = [1,2,4,6,7,8,9,9,9]
find_peak_mountain(arr)
````

# **In case of duplicate elements**
https://leetcode.com/problems/find-first-and-last-position-of-element-in-sorted-array/description/

We can find the first or the last occurence of the element:



````python
# first occurence
# first occurence = lower bound


def first_occurrence(arr, target):

    left, right = 0, len(arr)-1
    result = -1

    while left <= right:

        mid = (left + right) // 2

        if arr[mid] == target:
            result = mid
            right = mid - 1   # move left to confirm the first occurence, if any

        elif arr[mid] < target:
            left = mid + 1 # move right

        else:
            right = mid - 1 # move left

    return result

arr = [1,2,4,6,7,8,9,9]
target = 9

first_occurrence(arr, target)
````

````python
# last occurence
# upper bound - 1
## https://leetcode.com/problems/find-first-and-last-position-of-element-in-sorted-array/description/

def last_occurrence(arr, target):

    left, right = 0, len(arr)-1
    result = -1

    while left <= right:

        mid = (left + right) // 2

        if arr[mid] == target:
            result = mid
            left = mid + 1   # move right

        elif arr[mid] < target:
            left = mid + 1

        else:
            right = mid - 1

    return result

arr = [1,2,4,6,7,8,9,9]
target = 9

last_occurrence(arr, target)
````

````python
## Count Occurrences

# Use first + last occurrence

def count_occurrences(arr, target):

    first = first_occurrence(arr, target)
    last = last_occurrence(arr, target)

    if first == -1:
        return 0

    return last - first + 1

arr = [1,2,4,6,7,8,9,9,9,9]
target = 9

count_occurrences(arr, target)
````

# **Binary Search on Answer - SQRT(x)**
### Given a non-negative integer x, return the square root of x rounded down to the nearest integer. The returned integer should be non-negative as well.

https://leetcode.com/problems/sqrtx/description/

````python
def sqrt(x):
  l, r = 0, x
  ans = 1

  while l<=r:
    m = (l+r)//2
    if m*m > x:
      r = m - 1 # move left

    else:
      ans = m
      l = m + 1 # move right

  return ans


x = 10
sqrt(x)
````
