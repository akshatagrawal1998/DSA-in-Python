# **Bubble Sort**



### Bubble Sort is a sorting algorithm where:

→ We repeatedly compare adjacent elements

→ Swap them if they are in the wrong order

→ Largest elements “bubble up” to the end

Intuition: Imagine:

Numbers are like bubbles in water

→ Bigger bubble goes up

→ Smaller stays below

Example:

[5, 3, 2, 4]

After one pass:

[3, 2, 4, 5]
           ↑
     largest reached end


## Algorithm (Step-by-Step)
1. Start from index 0
2. Compare adjacent elements
3. Swap if left > right
4. Repeat till end
5. After each pass → largest element is fixed

```python
#bubble sort code:

def bubble_sort(arr):
  n = len(arr)

  for i in range(0, n):  # for iterations
    isswap = False
    for j in range(0, n-i-1):  #
      if arr[j] > arr[j+1]:
        #swap
        arr[j], arr[j+1] = arr[j+1], arr[j]
        isswap = True

    if isswap == False: # means no swapping in that iteration (means the array is now sorted)
      break


arr = [1,4,7,2,4,2,3]

bubble_sort(arr)
print(arr)
```

# **Insertion Sort**

### Take one element and insert it into the already sorted part.

### Usecase:
- Efficient for small datasets
- Works well when data is almost sorted

### Algorithm:

1. Start from second element
2. Compare with previous elements
3. Shift elements right, if not already sorted
4. Insert at correct position


## Complexity
Best: O(n) (already sorted)
Worst: O(n²)

```python
# insertion sort (ascending order)

def insertion_sort(arr):
  for i in range(1, len(arr)):
    key = arr[i]
    j = i-1

    while j >= 0 and arr[j] > key:
      arr[j+1] = arr[j]
      j = j - 1

    arr[j+1] = key



arr = [1,4,7,2,4,2,3]

insertion_sort(arr)
print(arr)
```

```python
# insertion sort (descending order)

def insertion_sort(arr):
  for i in range(1, len(arr)):
    key = arr[i]
    j = i-1

    while j >= 0 and arr[j] < key:
      arr[j+1] = arr[j]
      j = j - 1

    arr[j+1] = key



arr = [1,4,7,2,4,2,3]

insertion_sort(arr)
print(arr)
```

# **Selection Sort**

## Selection Sort works by:

→ Selecting the smallest element

→ Placing it at correct position

→ Repeating for remaining array

### Intuition: Instead of swapping again and again:

→ Find minimum

→ Put it in correct place

Visual Thinking

[5, 3, 2, 4]

Step 1 → smallest = 2 → put at index 0

### Algorithm

1. Start from index 0
2. Find smallest element in remaining array
3. Swap with current index if any smaller element is found
4. Move to next index

```python
# selection sort (ascending order)

def selection_sort(arr):

    n = len(arr)

    for i in range(n):

        min_index = i

        for j in range(i+1, n):

            if arr[j] < arr[min_index]:
                min_index = j

        arr[i], arr[min_index] = arr[min_index], arr[i]

    return arr
```

```python

arr = [1,4,7,2,4,2,3]

selection_sort(arr)
print(arr)
```

```python
# selection sort (descending order)

def selection_sort_desc(arr):

    n = len(arr)

    for i in range(n):

        min_index = i

        for j in range(i+1, n):

            if arr[j] > arr[min_index]:
                min_index = j

        arr[i], arr[min_index] = arr[min_index], arr[i]

    return arr
```

```python

arr = [1,4,7,2,4,2,3]

selection_sort_desc(arr)
print(arr)
```

```python
# merge sort code (ascending order)

def merge_sort(arr):
    if len(arr) <= 1:
        return arr

    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])

    # Added print statements to show left and right arrays after division
    print(f"Divided: Left: {left}, Right: {right}")

    return merge(left, right)



def merge(left, right):
    result = []
    i = j = 0

    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1


    # Add remaining elements
    # now when any one of the arrays (left or right) is finished, we attach the remaining part of the other array as it is without further comparision
    result.extend(left[i:])
    result.extend(right[j:])

    return result
```

```python

arr = [1,4,7,2,4,2,3]

sorted_array = merge_sort(arr)

print(sorted_array)
```

# **Quick Sort**

## Algorithm Steps:
1. Choose a pivot
2. Partition array:
3. Smaller elements → left
4. Larger elements → right
5. Recursively apply Quick Sort on left & right
6. Combine result


## Time Complexity
Best : O(n log n)
Average : O(n log n)
Worst :	O(n²)

### When Worst Case Happens?
When pivot is always smallest OR largest element

```python
def quick_sort(arr):
    # Base condition
    if len(arr) <= 1:
        return arr

    pivot = arr[len(arr) // 2]  # choose middle element

    left = []
    middle = []
    right = []

    for x in arr:
        if x < pivot:
            left.append(x)
        elif x == pivot:
            middle.append(x)
        else:
            right.append(x)

    return quick_sort(left) + middle + quick_sort(right)


arr = [7, 2, 5, 1, 9]
print(quick_sort(arr))
```
