**Singly Linked Lists** (SLL) and **Doubly Linked Lists** (DLL) are linear data structures where nodes are connected via pointers. SLL nodes contain data and a next pointer, allowing only forward traversal. DLL nodes contain data, next, and prev pointers, enabling bidirectional traversal. SLL is more memory-efficient, while DLL allows faster deletions.

A **circular linked list** is like a singly or doubly linked list with the first node, the "head", and the last node, the "tail", connected.

In singly or doubly linked lists, we can find the start and end of a list by just checking if the links are null. But for circular linked lists, more complex code is needed to explicitly check for start and end nodes in certain applications.

Circular linked lists are good for lists you need to cycle through continuously


### **Key Differences and Comparisons:**
##Structure:
SLL nodes point only to the next node. DLL nodes point to both next and previous nodes.

##Memory Usage:
SLL uses less memory because it has fewer pointers (2 fields: data, next). DLL uses more memory (3 fields: data, next, prev).

##Traversal:
SLL only allows forward traversal. DLL allows both forward and backward traversal.

##Deletion:
Deleting a node in an SLL takes O(n) time because finding the previous node requires traversal. In a DLL, deletion is O(1) if the node pointer is known, as the previous node is accessible.
Complexity: SLL is simpler to implement. DLL is more complex but more versatile, commonly used for implementing deques (double-ended queues).

- https://leetcode.com/problems/middle-of-the-linked-list/

Given the head of a singly linked list, return the middle node of the linked list.

If there are two middle nodes, return the second middle node.

```python
# basic approach - find the length of the linked list and iterate till half to get the mid element

# creating a linked list class again and a new linked list

class LinkedList:
  def __init__(self, data):
    self.data = data
    self.next = None

a = LinkedList(5)
b = LinkedList(40)
c = LinkedList(23)
d = LinkedList(512)
e = LinkedList(7)
f = LinkedList(89)

a.next = b
b.next = c
c.next = d
d.next = e
e.next = f

#traverse and printing the linked list
head = a
curr = head
len = 0

while curr!= None:
  print(curr.data, end = " -> ")
  curr= curr.next
  len+=1 # to calc the length of the linked list

print(f"\n\nLinked list has {len} elements")

curr = head
for i in range(len//2):
  curr = curr.next

print(f"\nMiddle element of the linked list is {curr.data}")
```

## Above solution is great but we have to iterate the list 2 times. Let's try to optimize it using the concept of fast & slow pointer.

## **Fast & Slow Pointer Concept**
### Here we take 2 pointers from begining and iterate both of them
- slow pointer as 1 step at a time (curr.next)
- fast pointer as 2 step at a time (curr.next.next)

### By this once the fast pointer reaches the end, the slow pointer will be at the mid as it's speed is half as of the fast pointer.

```python
 # Fast & Slow Pointer Concept - finding the mid element
 # alternate solution to the above question



head = a
slow = head
fast = head

while fast!=None and fast.next!=None: #(means we have reached the end of the list of 2nd last element)
  slow = slow.next
  fast = fast.next.next


print(f"\n\nLinked list has {len} elements")

print(f"\nMiddle element of the linked list is {slow.data}")

# here we need only 1 loop to traverse so time complexity is O(n)
```

# **Remove Duplicates from Sorted List II**

- https://leetcode.com/problems/remove-duplicates-from-sorted-list-ii/description/

# **Sort the Linked List**

we cannot directly access middle elements of linked list using index like arr[mid].

So algorithms like Quick Sort / Heap Sort are less practical. So, merge sort.

Time Complexity: O(n log n)
Space Complexity: O(n log n)

## Algorithm Steps: We do 3 things:

### **Step 1: Find Middle**
Use slow and fast pointer

slow moves 1 step, fast moves 2 steps

When fast reaches end → slow reaches middle.

### **Step 2: Split into two halves**

Break list:

slow.next = None

Now:

Left half (from head to mid) and Right half (from mid to end)

### **Step 3: Sort recursively**

Call same function again.

### **Step 4: Merge two sorted lists**

Like:

2 -> 4

1 -> 3

Compare nodes one by one:

1 smaller

then 2

then 3

then 4

Final:

1 -> 2 -> 3 -> 4

```python
# Definition for singly-linked list node
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


class Solution:
    def divide(self, head):
        # Base case: empty list or single node (already sorted)
        if not head or not head.next:
            return head

        # Step 1: Find middle of list
        slow, fast = head, head.next
        while fast and fast.next:
            slow = slow.next
            fast = fast.next.next

        mid = slow.next
        slow.next = None  # Split list into two halves

        # Step 2: Recursively sort both halves
        left = self.divide(head)
        right = self.divide(mid)

        # Step 3: Merge sorted halves
        return self.merge(left, right)

    def merge(self, l1, l2):
        dummy = ListNode()
        tail = dummy

        while l1 and l2:
            if l1.val < l2.val:
                tail.next = l1
                l1 = l1.next
            else:
                tail.next = l2
                l2 = l2.next
            tail = tail.next

        tail.next = l1 if l1 else l2
        return dummy.next
```

```python
# rotation logic

def rotateRight(head, k):
  p = head

  if head == None or head.next == None:  # single or 1 node in Linked list, so we return head only
    return head

  l = 0 # to calculate the length of the linked list

  while p.next!=None:
    p = p.next
    l = l + 1

  l = l+1

  k = k%l
  if k == 0:
    return head

  else:
    curr = head

    for i in range(l-k-1):
      curr = curr.next

    p.next = head # connecting the last node to the head node
    head = curr.next # making the current's next node as curr next
    curr.next = None

    return head


printlinkedlist(rotateRight(head, 3))
```

# **Intersection of Two Linked List**

- https://leetcode.com/problems/intersection-of-two-linked-lists/

Given the heads of two singly linked-lists headA and headB, return the node at which the two lists intersect. If the two linked lists have no intersection at all, return null.

Screenshot 2026-04-30 at 4.36.57 PM.png


#**Solution**

## Approach 1:
Given head1, head 2 -> find the length of both the linked lists (l1, l2)
let's say l1 = 5, l2 = 6. move the first list by (6-5)= 1 position.
Now keep iterating both the lists 1 by 1 and when both the pointer reaches to a common position, that's the intersection point.


## Approach 2:
Intuition: Both moves with same speed (one step at a time) and when they cover same distance, they'll meet at a common point and that's the intersection point.

Use two pointers:

p1 starts from headA
p2 starts from headB

When one pointer reaches end, move it to the other list's head.

Eventually:

If intersection exists → they meet there
If no intersection → both become null



# Time Complexity: O(m+n)

# Space Complexity: O(1)

```python
# Definition for singly-linked list
class ListNode:
    def __init__(self, val=0):
        self.val = val
        self.next = None


# Create Intersecting Linked Lists

# Common Part (intersection starts here)
common = ListNode(8)
common.next = ListNode(4)
common.next.next = ListNode(5)

# List A = 4 -> 1 -> 8 -> 4 -> 5
headA = ListNode(4)
headA.next = ListNode(1)
headA.next.next = common

# List B = 5 -> 6 -> 1 -> 8 -> 4 -> 5
headB = ListNode(5)
headB.next = ListNode(6)
headB.next.next = ListNode(1)
headB.next.next.next = common
```

```python
# Solution: Find Intersection Node


def getIntersectionNode(headA, headB):
    p1 = headA
    p2 = headB
    c = 0

    while True:
      if p1==p2:  # we have found the common value between them
        return p1

      p1 = p1.next # if not found the common point, keep moving ahead
      p2 = p2.next

      if p2==None: # if p2 pointer has covered the whole list B, start from headA
        p2 = headA
        c = c + 1 # counter variable if any pointer has completed the list

      if p1==None:
        p1 = headB
        c = c + 1  # counter variable if any pointer has completed the list

      if c > 1:  # c>1 means the list has been traversed more than twice and if intersection not found. So maybe intersection does not exist.
        return None



result = getIntersectionNode(headA, headB)

if result:
    print(result.val)
else:
    print(None)
```

# **Add Two Numbers**

- https://leetcode.com/problems/add-two-numbers/


You are given two linked lists.

Each linked list stores a number digit by digit in reverse order.

That means:

[2,4,3] = 342
[5,6,4] = 465

Because:

first node = ones place
second node = tens place
third node = hundreds place

Now add both numbers:

342 + 465 = 807

Return answer as linked list:

[7,0,8]

Because reverse order of 807 is:

7 -> 0 -> 8


# Algorithm:

###Step 1: Create dummy node for result list.

###Step 2: Keep carry = 0

###Step 3: Loop until:
l1 exists OR l2 exists OR carry exists

###Step 4: Take values until l1 or l2 has values
x = l1.val if l1 else 0

y = l2.val if l2 else 0

###Step 5: Find total:
total = x + y + carry

### Step 6: New digit:

digit = total % 10

carry = total // 10

###Step 7: Create node with digit.

###Step 8: Move l1 and l2 forward.

###Step 9: Return result.


##Why Dummy Node?

Instead of handling first node separately.

dummy = ListNode(0)
current = dummy

At end:

return dummy.next

Very common linked list trick.

```python
# singly-linked list node - creation for the above question

class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


# Function to create linked list from Python list
def create_linked_list(arr):
    dummy = ListNode(0)      # temporary dummy node
    current = dummy

    for num in arr:
        current.next = ListNode(num)
        current = current.next

    return dummy.next       # actual head


# Function to print linked list
def print_linked_list(head):
    while head:
        print(head.val, end=" -> " if head.next else "")
        head = head.next
    print()


# Create l1 = [2,4,3]
l1 = create_linked_list([2, 4, 3])

# Create l2 = [5,6,4]
l2 = create_linked_list([5, 6, 4])


# Check lists
print("l1:")
print_linked_list(l1)

print("l2:")
print_linked_list(l2)
```

```python
class Solution:
    def addTwoNumbers(self, l1, l2):

        # Dummy node to start result list
        dummy = ListNode(0)

        # Current pointer for result list
        current = dummy

        # Carry value (like school addition)
        carry = 0

        # Loop until both lists end and carry becomes 0
        while l1 or l2 or carry:

            # Take value from l1, else 0
            x = l1.val if l1 else 0

            # Take value from l2, else 0
            y = l2.val if l2 else 0

            # Add both digits + carry
            total = x + y + carry

            # New carry
            carry = total // 10

            # Digit to store in node
            digit = total % 10

            # Create new node
            current.next = ListNode(digit)

            # Move result pointer
            current = current.next

            # Move l1 ahead if exists
            if l1:
                l1 = l1.next

            # Move l2 ahead if exists
            if l2:
                l2 = l2.next

        # Return result list
        return dummy.next



# Create input linked lists
l1 = create_linked_list([2, 4, 3])
l2 = create_linked_list([5, 6, 4])


print("l1:")
print_linked_list(l1)

print("l2:")
print_linked_list(l2)

# Call function
obj = Solution()
result = obj.addTwoNumbers(l1, l2)

# Print result
print("\n\nOutput:")
print_linked_list(result)
```

# **Linked List Cycle**

- https://leetcode.com/problems/linked-list-cycle/

we are given the head of a linked list.

We need to check:

## Does this linked list contain a cycle (loop)?

A cycle means: Instead of ending with None, one node points back to a previous node.

So traversal becomes infinite.

## Solution Approach: Fast & Slow Pointer (Floyd's Algorithm)

slow pointer → moves 1 step
fast pointer → moves 2 steps

If cycle exists:

- Fast pointer will eventually catch slow pointer.

If no cycle:

- Fast pointer reaches None

This uses O(1) memory.


## Algorithm:
###Step 1: If list is empty → return False

###Step 2: Create two pointers:
slow = head
fast = head

###Step 3: Loop while:
fast and fast.next

### Step 4: Move:

slow = slow.next

fast = fast.next.next

###Step 5:

If both meet:

if slow == fast:
    return True (means cycle exists)

###Step 6: If loop ends -> return False


## Time Complexity: O(n) - At most traverse list once.

## Space Complexity: O(1) - Only two pointers used.

```python
# Definition for singly-linked list
class ListNode:
    def __init__(self, val=0):
        self.val = val
        self.next = None

# Create nodes
node1 = ListNode(3)
node2 = ListNode(2)
node3 = ListNode(0)
node4 = ListNode(-4)

# Connect nodes
node1.next = node2
node2.next = node3
node3.next = node4

# Create cycle: last node points to node2
node4.next = node2

head = node1
```

```python
class Solution:
    def hasCycle(self, head):

        # Two pointers start from head
        slow = head
        fast = head

        # Run until fast reaches end
        while fast and fast.next:

            # Slow moves 1 step
            slow = slow.next

            # Fast moves 2 steps
            fast = fast.next.next

            # If both meet -> cycle exists
            if slow == fast:
                return True

        # If fast reached end -> no cycle
        return False

obj = Solution()

print(obj.hasCycle(head))
```

# **Linked List Cycle II**

- https://leetcode.com/problems/linked-list-cycle-ii/

You are given the head of a linked list.

You already know a cycle may exist.

Now instead of returning True/False, you must return:

The node where the cycle starts

If no cycle exists:

return None

## **Solution (Floyd’s Algorithm)**
Same as previous problem:

### Step 1: Detect Cycle:

slow = 1 step

fast = 2 steps

If they meet → cycle exists.

###Step 2:

Put one pointer at head.

Keep second pointer where meeting happened.

Move both 1 step each.

Where they meet again = cycle start


#Why This Works?

Distance math guarantees:

head -> cycle start
=
meeting point -> cycle start

So walking both together makes them meet exactly at cycle start.

## Full Algorithm
###Step 1:
slow = head
fast = head

###Step 2: Move pointers:

slow = slow.next
fast = fast.next.next

Until they meet.

###Step 3: If no meet -> return None (no cycle exists)

###Step 4: Create new pointer:

ptr = head

###Step 5: Move both 1 step:

ptr = ptr.next
slow = slow.next

###Step 6: When equal: -> return ptr

That node is cycle start.

```python
class ListNode:
    def __init__(self, val=0):
        self.val = val
        self.next = None


# Create nodes
node1 = ListNode(3)
node2 = ListNode(2)
node3 = ListNode(0)
node4 = ListNode(-4)

# Connect nodes
node1.next = node2
node2.next = node3
node3.next = node4

# Create cycle
node4.next = node2

head = node1
```

```python
class Solution:
  def detectCycle(self, head):

          slow = head
          fast = head

          # Step 1: Detect cycle
          while fast and fast.next:

              slow = slow.next
              fast = fast.next.next

              # Meeting point found
              if slow == fast:

                  # Step 2: Find cycle start
                  ptr = head

                  while ptr != slow:
                      ptr = ptr.next
                      slow = slow.next

                  return ptr

          # No cycle
          return None
```

```python
obj = Solution()

result = obj.detectCycle(head)

print(result.val if result else None)
```
