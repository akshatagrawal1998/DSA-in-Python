````python
# Stack Class Implementation
class Stack:
    def __init__(self):
        self.items = []

    def push(self, value):
        self.items.append(value)

    def pop(self):
        if self.is_empty():
            return "Stack Underflow"
        return self.items.pop()

    def peek(self):
        if self.is_empty():
            return None
        return self.items[-1]

    def is_empty(self):
        return len(self.items) == 0

    def size(self):
        return len(self.items)

    def display(self):
        print(self.items)


s = Stack()

s.push(5)
s.push(100)
s.push(15)

s.display()

print(s.pop())
s.display()

print(s.peek())
````

# Stack using Linked List

##Instead of array/list:

We store elements as nodes:

Top → [30] → [20] → [10] → None

Top = Head of Linked List

Insert/Delete happens at head

push(x)	Insert at head

pop()	Delete head

peek()	Return head value

````python
# Node class for Linked List
class Node:
    def __init__(self, data):
        self.data = data      # value stored in node
        self.next = None      # pointer to next node


# Stack class using Linked List
class Stack:
    def __init__(self):
        self.top = None       # top points to head node
        self._size = 0        # to track number of elements

    # PUSH operation (insert at beginning)
    def push(self, value):
        new_node = Node(value)      # create new node

        new_node.next = self.top    # new node points to current top

        self.top = new_node         # update top to new node

                    # increase size

    # POP operation (remove from beginning)
    def pop(self):
        if self.is_empty():
            print("Stack Underflow")
            return None

        removed_value = self.top.data   # store value to return

        self.top = self.top.next        # move top pointer to next node

        self._size -= 1

        return removed_value

    # PEEK operation (see top element)
    def peek(self):
        if self.is_empty():
            return None
        return self.top.data

    # Check if stack is empty
    def is_empty(self):
        return self.top is None

    # Return size
    def size(self):
        return self._size

    # Display stack
    def display(self):
        temp = self.top
        while temp:
            print(temp.data, end=" -> ")
            temp = temp.next
        print("None")
````

````python
s = Stack()

# Push elements
s.push(10)
s.push(20)
s.push(30)

print("Stack:")
s.display()

# Pop
print("Popped:", s.pop())

# Peek
print("Top element:", s.peek())

# Size
print("Size:", s.size())
````

````python
# Queue class implementation using Python List (not prod used)

class Queue:
    def __init__(self):
        self.q = []          # list to store elements
        self.front = -1      # front pointer

    # Enqueue operation (insert at rear)
    def push(self, x):
        # If queue is empty, initialize front
        if self.front == -1:
            self.front = 0

        self.q.append(x)     # add element at end

    # Dequeue operation (remove from front)
    def pop(self):
        # If queue is empty
        if self.front == -1 or self.front >= len(self.q):
            print("Queue Underflow")
            return -1

        x = self.q[self.front]   # get front element
        self.front += 1          # move front pointer

        return x

    # Get front element
    def top(self):
        if self.front == -1 or self.front >= len(self.q):
            return -1
        return self.q[self.front]

    # Get size of queue
    def size(self):
        return len(self.q) - self.front if self.front != -1 else 0

    # Check if queue is empty
    def is_empty(self):
        return self.size() == 0

    # Display queue (only valid elements)
    def display(self):
        if self.is_empty():
            print("Queue is empty")
            return

        for i in range(self.front, len(self.q)):
            print(self.q[i], end=" -> ")
        print("None")
````

````python
q = Queue()

q.push(10)
q.push(20)
q.push(30)

print("Queue:")
q.display()

print("Removed:", q.pop())
print("Front:", q.top())

q.display()

print("Size:", q.size())
````

````python
# queue class implementation

from collections import deque

class Queue:
    def __init__(self):
        self.items = deque()

    def enqueue(self, value):
        self.items.append(value)

    def dequeue(self):
        if self.is_empty():
            return "Queue Empty"
        return self.items.popleft()

    def front(self):
        if self.is_empty():
            return None
        return self.items[0]

    def rear(self):
        if self.is_empty():
            return None
        return self.items[-1]

    def is_empty(self):
        return len(self.items) == 0

    def size(self):
        return len(self.items)
````

# **Queue using Linked List**
### Using a normal list for queue can be inefficient:

list.pop(0)  # O(n) time complexity because shifting happens.

Linked List + two pointers (front & rear) gives:

Enqueue = O(1) time complexity
Dequeue = O(1) time complexity


### Queue follows: FIFO = First In First Out

We maintain:

Front → first element (removal)

Rear  → last element (insertion)


### Appplications:
Real-World Use Cases
1. Call Center Queue: Customers handled in order.

2. Printer Queue: Jobs printed in order.

3. CPU Scheduling: Tasks executed FIFO.

4. BFS in Graph: Level-wise traversal.

````python
# Node class for Linked List
class Node:
    def __init__(self, data):
        self.data = data      # value stored in node
        self.next = None      # pointer to next node


# Queue class using Linked List
class Queue:
    def __init__(self):
        self.front = None     # points to first node
        self.rear = None      # points to last node
        self._size = 0        # track size

    # ENQUEUE operation (insert at rear)
    def enqueue(self, value):
        new_node = Node(value)   # create new node

        # If queue is empty
        if self.rear is None:
            self.front = self.rear = new_node
        else:
            self.rear.next = new_node  # link last node to new node
            self.rear = new_node       # update rear pointer

        self._size += 1

    # DEQUEUE operation (remove from front)
    def dequeue(self):
        if self.is_empty():
            print("Queue Underflow")
            return None

        removed_value = self.front.data

        self.front = self.front.next   # move front pointer

        # If queue becomes empty
        if self.front is None:
            self.rear = None

        self._size -= 1

        return removed_value

    # Get front element
    def get_front(self):
        if self.is_empty():
            return None
        return self.front.data

    # Get rear element
    def get_rear(self):
        if self.is_empty():
            return None
        return self.rear.data

    # Check empty
    def is_empty(self):
        return self.front is None

    # Size
    def size(self):
        return self._size

    # Display queue
    def display(self):
        temp = self.front
        while temp:
            print(temp.data, end=" -> ")
            temp = temp.next
        print("None")
````

````python
q = Queue()

# Enqueue
q.enqueue(10)
q.enqueue(20)
q.enqueue(30)

print("Queue:")
q.display()

# Dequeue
print("Removed:", q.dequeue())

# Front & Rear
print("Front:", q.get_front())
print("Rear:", q.get_rear())

# Size
print("Size:", q.size())



# all these operation are O(1) time complexity


````

Let’s say array size = 5

[_, _, 30, _, _]

Rear moves → 3 → 4 → ❌ FULL

Even though space exists on left!

Wasted space (false overflow)

Memory is not reused → inefficient

Reuses empty spaces

## Queue using Two Stacks — Intuition & Algorithm

A **queue** is like a line at a ticket counter — whoever joins first, leaves first (**FIFO**). A **stack** is like a pile of plates — whatever you put on top, comes off first (**LIFO**). These are opposites. So how do we fake a queue using stacks?

> **Key Insight:** If you pour a stack into another stack, the order flips. Pour it again, it flips back. One flip is all we need — it turns LIFO into FIFO.

Think of it like a tube of tennis balls. You drop balls in from the top (stack 1). To get the bottom ball out, you tip them all into another tube (stack 2) — now the bottom ball is on top. That's the entire trick.

---

### The Two Stacks and Their Jobs

| Stack | Name | Only job |
|---|---|---|
| Stack 1 | `stack_in` | Accepts every new item that arrives |
| Stack 2 | `stack_out` | Serves every item that leaves |

They never mix roles. `stack_in` never pops. `stack_out` never receives a direct push.


### Operation 1 — push(x)

**Goal:** Add a new item to the back of the queue.

**Intuition:** The back of the queue is always `stack_in`. New arrivals just pile on top of it. We don't care about order yet — that's a problem for later.

**Algorithm:**
```
1. Take the new item x
2. Push x onto the top of stack_in
3. Done. No other stack is touched.
```

**Why it's O(1):** One action. Always. No matter how many items exist.

### Operation 2 — pop()

**Goal:** Remove and return the item that has been waiting the longest (the front of the queue).

**Intuition:** The front of the queue is the item that was pushed first — which means it's buried at the bottom of `stack_in`. We can't reach the bottom of a stack directly. So we pour everything into `stack_out`, which flips the order. Now that oldest item is sitting right on top of `stack_out`, ready to be popped immediately.

**The golden rule:** Only pour `stack_in` into `stack_out` when `stack_out` is completely empty. If `stack_out` still has items, they are already in the right order — use them first.

**Algorithm:**
```
1. If both stacks are empty → error, queue is empty, stop.

2. Check if stack_out is empty.
   If YES (stack_out is empty):
       While stack_in is not empty:
           Pop the top item from stack_in
           Push that item onto stack_out
       (All items are now in stack_out, order is reversed)
   If NO (stack_out already has items):
       Skip the transfer entirely — items are already ready

3. Pop the top item from stack_out
4. Return that item — this is the front of the queue
```

**Why it's O(1) amortized:** A single pop can trigger an O(n) transfer. But each item is only ever transferred once in its entire lifetime — pushed into `stack_in` once, moved to `stack_out` once, popped once. So the total cost spread across all operations averages out to O(1) per operation.

### Operation 3 — peek()

**Goal:** Look at the front item without removing it.

**Intuition:** Identical thinking to pop — we need the oldest item at the top of `stack_out`. The only difference is that after we find it, we leave it there.

**Algorithm:**
```
1. If both stacks are empty → error, queue is empty, stop.

2. Check if stack_out is empty.
   If YES:
       Transfer everything from stack_in into stack_out
       (same transfer loop as pop)
   If NO:
       Skip transfer

3. Read the top item of stack_out (do NOT remove it)
4. Return that item
```

**Key difference from pop:** Step 3 reads the top, not removes it. The item stays in `stack_out` for the next pop or peek call.

### Operation 4 — empty()

**Goal:** Report whether the queue has any items at all.

**Intuition:** Items live in one of two places — either waiting in `stack_in`, or ready in `stack_out`. If both are empty, the queue is empty. Even one item in either stack means the queue is not empty.

**Algorithm:**
```
1. Check if stack_in has zero items AND stack_out has zero items
2. If both are zero → return True (queue is empty)
3. If either has items → return False (queue is not empty)
```

**Why it's O(1):** Just checking two lengths. No movement of items.

### The Transfer — Visualised Step by Step

Say you pushed 1, 2, 3 in that order. Here is the state before a pop:

```
stack_in          stack_out
┌───┐             ┌───┐
│ 3 │ ← top       │   │  (empty)
│ 2 │             │   │
│ 1 │ ← bottom    │   │
└───┘             └───┘
Front of queue = 1, but it's buried!
```

Transfer fires (because `stack_out` is empty):

```
stack_in          stack_out
┌───┐             ┌───┐
│   │  (empty)    │ 1 │ ← top   ✓ front of queue is now on top!
│   │             │ 2 │
│   │             │ 3 │ ← bottom
└───┘             └───┘
```

Now pop() simply removes 1 from the top of `stack_out`. Next pop gets 2. Then 3. No new transfer is needed until `stack_out` runs dry again.

### The One Rule to Remember

> **Transfer from `stack_in` to `stack_out` only when `stack_out` is empty — never otherwise.**

Breaking this rule destroys the queue order. Following it guarantees correct FIFO behaviour every single time, regardless of how many pushes and pops are mixed together.

````python
class MyQueue:

    def __init__(self):
        self.in_stack = []
        self.out_stack = []

    def push(self, x: int) -> None:
        self.in_stack.append(x)

    def pop(self) -> int:
        self._move()
        return self.out_stack.pop()

    def peek(self) -> int:
        self._move()
        return self.out_stack[-1]

    def empty(self) -> bool:
        return not self.in_stack and not self.out_stack

    def _move(self):
        if not self.out_stack:
            while self.in_stack:
                self.out_stack.append(self.in_stack.pop())
````

````python
q = MyQueue()

q.push(10)
q.push(20)
q.push(30)

print(q.peek())   # Should print 10
print(q.pop())    # Should print 10
print(q.pop())    # Should print 20
print(q.empty())  # Should print False
print(q.pop())    # Should print 30
print(q.empty())  # Should print True
````

# Strategy
Push element into queue

Rotate previous elements behind it

## Why this works?

Because queue front should always behave like stack top

## Algorithm
push(x): Add element

Rotate queue (size - 1 times)

pop(): Remove front

top(): Return front

## “After every push, make sure the newest element comes to the front of the queue.”

# Time Complexity:
push -> O(n)

pop -> O(1)

top -> O(1)

empty -> O(1)

````python
from collections import deque

class MyStack:

    def __init__(self):
        self.q = deque()

    def push(self, x: int) -> None:
        self.q.append(x)

        # Rotate elements
        for _ in range(len(self.q) - 1):
            self.q.append(self.q.popleft())

    def pop(self) -> int:
        return self.q.popleft()

    def top(self) -> int:
        return self.q[0]

    def empty(self) -> bool:
        return len(self.q) == 0
````

````python
# MAPPING APPROACH USING DICTIONARY

def isValid(s: str):
    stack = []

    mapping = {
        ')': '(',
        ']': '[',
        '}': '{'
    }

    for char in s:
        if char in mapping:  # closing bracket
            if not stack:
                return False
# s = '[]]{[]}'
            top = stack.pop()
            if top != mapping[char]:
                return False
        else:
            # opening bracket
            stack.append(char)

    return len(stack) == 0
````

````python
s = "{[]}"
isValid(s)


# Time & Space Complexity
# Time	O(n)
# pace	O(n)
````

````python
# alternate approach

def isValid(s):
    st = []

    for ch in s:
        if ch in "({[":
            st.append(ch)
        else:
            if not st:   # if len(st) == 0: True if st is empty
                return False

            top = st.pop()

            if ch == ')' and top != '(':
                return False
            elif ch == '}' and top != '{':
                return False
            elif ch == ']' and top != '[':
                return False

    return len(st) == 0


s = "{[]}"
isValid(s)
````

# **Browser History System**

This is a **very real industry problem**—exactly how browsers like Google Chrome or Mozilla Firefox manage navigation.

Let’s build it using **system design + DSA**.

# 1. Problem Understanding

When we use a browser:

* we **visit a new page**
* we can go **back**
* we can also go **forward**

So internally, the system needs to track:

* Past pages
* Current page
* Future pages (after going back)

# 2. Core Idea (Industry-Level Design)

We use **2 stacks**:

###back_stack: Stores history (pages behind current)

### forward_stack: Stores pages ahead (when user presses back)

### current: Stores current page

# 3. Intuition with Example

### Steps:

```
visit google.com
visit youtube.com
visit github.com
```

State:

```
back_stack   = [google, youtube]
current      = github
forward_stack = []
```

---

### Press BACK ⬅️

```
forward_stack = [github]
current = youtube
back_stack = [google]
```

---

### Press BACK ⬅️ again

```
forward_stack = [github, youtube]
current = google
back_stack = []
```

---

### Press FORWARD ➡️

```
back_stack = [google]
current = youtube
forward_stack = [github]
```

---

# 4. Algorithm

## visit(url)

* Push current → back_stack
* Update current
* Clear forward_stack (important!)

## back()

* If back_stack empty → stay
* Push current → forward_stack
* Pop from back_stack → current

## forward()

* If forward_stack empty → stay
* Push current → back_stack
* Pop from forward_stack → current


# 7. Time Complexity

| Operation | Complexity |
| --------- | ---------- |
| visit     | O(1)       |
| back      | O(1)       |
| forward   | O(1)       |


# 8. Why This Works?

Stacks naturally support:

* “Last visited → first to go back”
* “Undo / redo” behavior

This is the same concept used in:

* Text editors (Undo/Redo)
* IDE navigation
* Mobile app navigation


# 9. Real-World Engineering:

In real systems (like Google Chrome):

They don’t just store URLs, they store:

* Page state (scroll position, form data)
* Cache snapshots
* Memory optimizations (limit stack size)



````python
class BrowserHistory:

    def __init__(self, homepage):
        self.current = homepage
        self.back_stack = []
        self.forward_stack = []

    def visit(self, url):
        self.back_stack.append(self.current)
        self.current = url
        self.forward_stack.clear()  # Important

    def back(self):
        if not self.back_stack:
            return self.current

        self.forward_stack.append(self.current)
        self.current = self.back_stack.pop()
        return self.current

    def forward(self):
        if not self.forward_stack:
            return self.current

        self.back_stack.append(self.current)
        self.current = self.forward_stack.pop()
        return self.current
````

````python

browser = BrowserHistory("google.com")

browser.visit("youtube.com")
browser.back()
````

#**Task Scheduling**
A printer receives tasks:

Print A

Print B

Print C

### Hint: Use Queue to process tasks one by one.

````python
from collections import deque

class PrinterQueue:

    def __init__(self):
        self.queue = deque()

    def add_task(self, task):
        self.queue.append(task)
        print(f"Task added: {task}")

    def process_task(self):
        if not self.queue:
            print("No tasks to process")
            return

        task = self.queue.popleft()
        print(f"Processing: {task}")

    def is_empty(self):
        return len(self.queue) == 0
````

````python
pq = PrinterQueue()

pq.add_task("Print A")
pq.add_task("Print B")
pq.add_task("Print C")

pq.process_task()
pq.process_task()
pq.process_task()
````
